## Análise Exploratória de Dados (EDA).

```python
# Qual região possui maior faturamento?
faturamento = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
plt.figure(figsize=(10, 6))
sns.barplot(x=faturamento.index, y=faturamento.values)
plt.title('Faturamento por Região')
barras = plt.bar(faturamento.index, faturamento.values)
plt.bar_label(barras, labels=[f'R${v:,.2f}' for v in faturamento.values], padding=3)
plt.xlabel('Região')
plt.ylabel('Faturamento')
plt.show()  
```

A região West lidera o faturamento da empresa, representando o principal mercado consumidor. A região South apresenta o menor volume de vendas, indicando potencial para expansão comercial ou necessidade de revisão da estratégia de atuação.

```python
# Qual Estado vende mais?
estado_vendas = df.groupby("State")['Sales'].sum().sort_values(ascending=False).head(5)
plt.figure(figsize=(12, 6))
sns.barplot(x=estado_vendas.index, y=estado_vendas.values)
plt.title("Top 5 estados com maior Vendas")
barras = plt.bar(estado_vendas.index, estado_vendas.values)
plt.bar_label(barras, labels=[f'R${v:,.2f}' for v in estado_vendas.values], padding=1)
plt.xlabel("Estado")
plt.ylabel("Vendas")    
plt.show()
```

A Califórnia concentra aproximadamente 47% mais faturamento que Nova York, demonstrando elevada participação nas vendas totais.

```python
# Qual Segmento compra mais?
segmento_vendas = df.groupby('Segment')["Sales"].sum().sort_values(ascending=False)
plt.figure(figsize=(12, 6))
sns.barplot(x=segmento_vendas.index, y=segmento_vendas.values)
plt.title("Segmentos por maior Vendas")
barras = plt.bar(segmento_vendas.index, segmento_vendas.values)
plt.bar_label(barras, labels=[f'R${v:,.2f}' for v in segmento_vendas.values], padding=1)
plt.xlabel("Segmento")
plt.ylabel("Vendas")    
plt.show()
```

O segmento Consumer representa mais da metade do faturamento da empresa

```python
# categoria que mais vende?
categoria_vendas = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
plt.figure(figsize=(12,6))
plt.title("Categorias por maior Vendas")
plt.bar(categoria_vendas.index, categoria_vendas.values)
barras = plt.bar(categoria_vendas.index, categoria_vendas.values)
plt.bar_label(barras, labels=[f"R${v:,.2f}" for v in categoria_vendas.values], padding=1)
plt.ylabel("Vendas")
plt.xlabel("Categoria")
plt.show()
```

Embora a diferença entre as categorias não seja muito grande, Technology apresenta o maior faturamento.

```python
subcategoria_vendas = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(12,6))
plt.title("Subcategoria por maior vendas")
plt.bar(subcategoria_vendas.index, subcategoria_vendas.values)
barras = plt.bar(subcategoria_vendas.index, subcategoria_vendas.values)
plt.bar_label(barras, labels=[f'${v:,.2f}' for v in subcategoria_vendas.values], padding=1)
plt.ylabel("Vendas")
plt.xlabel("Subcategoria")
plt.show()
```

O faturamento está relativamente distribuído entre várias subcategorias.
