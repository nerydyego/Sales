```python
# Variancia 
df[['Sales', 'Quantity', 'Discount', 'Profit']].var()
```

### Conclusão da variância
* A variância confirma que Sales e Profit são as variáveis mais heterogêneas do conjunto de dados, apresentando ampla dispersão e grande variabilidade entre as observações. Já Quantity e Discount possuem menor dispersão, indicando um comportamento mais homogêneo.

```python
# Assimetria
df[['Sales', 'Quantity', 'Discount', 'Profit']].skew()
```

```python
df["Discount"].value_counts(normalize=True).sort_index()
```

### Conclusão da assimetria
Todas as variáveis apresentam assimetria positiva, indicando predominância de observações concentradas em valores baixos e uma cauda estendida para valores mais elevados. As variáveis Sales e Profit exibem assimetria extremamente elevada, evidenciando a influência de poucos registros com valores muito acima da maioria.

```python
# Curtose
df[['Sales', 'Quantity', 'Discount', 'Profit']].kurt()
```

### Conclusão da curtose

As variáveis Sales e Profit apresentam curtose extremamente elevada, indicando distribuições leptocúrticas, caracterizadas por forte concentração de observações próximas ao centro e pela presença de caudas pesadas. Esse comportamento reforça a existência de valores extremos já identificados pelos boxplots e histogramas.

```python
# Histograma das variáveis numéricas
plt.figure(figsize=(10, 6))
sns.histplot(
    np.log1p(df['Sales']),
    kde=True,
    bins=30
)
plt.title('Distribuição de Sales')
plt.show()

#A função np.log1p() aplica log(1 + x), permitindo visualizar muito melhor a 
# estrutura dos dados quando há grande assimetria positiva.

'''
plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x='Sales',
    kde=True,
    bins=30
)
plt.title('Distribuição de Sales')
plt.show()
'''
```

### Histograma de Sales
* enorme concentração entre valores baixos;
* poucos pedidos muito altos;
* cauda longa até aproximadamente R$ 22.000.

indicando:
* forte assimetria positiva;
* presença de outliers;
* distribuição distante da normal.

```python
plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x='Profit',
    kde=True,
    bins=30
)
plt.title('Distribuição de Profit')
plt.show()
```

### Histograma Profit
* concentração próxima de zero;
* existência de prejuízos elevados;
* existência de lucros elevados;
* maior extensão da cauda para o lado positivo.

Esse gráfico mostra claramente que a empresa possui tanto operações muito lucrativas quanto operações que geram perdas significativas.

```python
plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x='Quantity',
    kde=True,
    bins=30
)
plt.title('Distribuição de Quantity')
plt.show()
```

### Histograma de Quantity
* maior concentração entre 2 e 5 unidades;
* redução gradual da frequência conforme a quantidade aumenta;
* poucos pedidos acima de 10 unidades.

É uma distribuição típica de quantidade comprada.

```python
plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x='Discount',
    kde=True,
    bins=30
)
plt.title('Distribuição de Discount')
plt.show()
```

### Histograma Discount
* *isso* não é uma distribuição contínua.
* Observe que existem "picos".

Isso ocorre porque o desconto assume poucos valores possíveis.
