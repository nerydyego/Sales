```python
features = [
    "Ship Mode",
    'Segment',
    'Region',
    'Category',
    'Sub-Category',
    'Sales',
    'Quantity',
    'Discount'
    ] 

target =  'Profit'
```

```python
X = df[features] # variáveis independentes/preditoras
y = df[target] #variável dependente/alvo
```

#### Importação de bibliotecas scikit-learn > train_test_split

```python
from sklearn.model_selection import train_test_split
```

```python
# 
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
X_train['Segment'].unique()
```

#### Criação do One_hot_encoder

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

features_numericas = [
    'Sales',
    'Quantity',
    'Discount'
]

features_categoricas = [
    'Ship Mode',
    'Segment',
    'Region',
    'Category',
    'Sub-Category'
]

encoder = OneHotEncoder(
    handle_unknown='ignore'
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            'cat',
            encoder,
            features_categoricas
        )
    ],
    remainder='passthrough'
)
```

```python
preprocessor
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

modelo_lr = LinearRegression()

pipeline_lr = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('modelo', modelo_lr)
    ]
)

pipeline_lr.fit(X_train, y_train)

y_pred_lr = pipeline_lr.predict(X_test)
```

```python
print(y_pred_lr[:10])
```

```python
comparacao = pd.DataFrame({
    'Profit Real': y_test.values,
    'Profit Previsto': y_pred_lr
})

comparacao.head(10)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae_lr = mean_absolute_error(y_test, y_pred_lr)
mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)

print(f"MAE:  {mae_lr:.2f}")
print(f"RMSE: {rmse_lr:.2f}")
print(f"R²:   {r2_lr:.4f}")

print()
print(f"Média Profit teste:   {y_test.mean():.2f}")
print(f"Mediana Profit teste: {y_test.median():.2f}")
print(f"Desvio padrão:        {y_test.std():.2f}")
```
