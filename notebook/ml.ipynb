#### Importação de bibliotecas scikit-learn > train_test_split

````python
from sklearn.model_selection import train_test_split
````

````python
# 
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
````

#### Criação do One_hot_encoder

````python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

features_numericas = [
    'Sales',
    'Quantity',
    'Discount'
]

features_categoricas = [
    'Ship Mode',
    'Segment',
    'Region',
    'Category',
    'Sub-Category'
]

encoder = OneHotEncoder(
    handle_unknown='ignore'
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            'cat',
            encoder,
            features_categoricas
        )
    ],
    remainder='passthrough'
)
````

````python
preprocessor
````

````python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

modelo_lr = LinearRegression()

pipeline_lr = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('modelo', modelo_lr)
    ]
)

pipeline_lr.fit(X_train, y_train)

y_pred_lr = pipeline_lr.predict(X_test)
````

````python
# Importação da biblioteca DecisionTreeRegressor do scikit-learn
from sklearn.tree import DecisionTreeRegressor
````

````python
# modelo sem poda explicita, avaliação de overfitting, sem regularização, sem poda, sem limite de profundidade, sem limite de folhas, sem limite de amostras mínimas por folha
modelo_dt = DecisionTreeRegressor(random_state=42)

#Criação do pipeline para o modelo de árvore de decisão
pipeline_dt = Pipeline(
    steps=[
        ('processor', preprocessor),
        ('modelo', modelo_dt)
    ]
)
````

````python
# Previsão dos valores de Profit para o conjunto de teste usando o modelo de árvore de decisão
y_pred_dt_train = pipeline_dt.predict(X_train)
y_pred_dt = pipeline_dt.predict(X_test)
# Métricas
metricas_dt_train = calcular_metricas_regressao(
    y_train,
    y_pred_dt_train
)
resultados = registrar_resultado(
    resultados,
    "Decision Tree _ train",
    metricas_dt_train
)
metricas_dt = calcular_metricas_regressao(
    y_test,
    y_pred_dt
)
resultados = registrar_resultado(
    resultados,
    "Decision Tree test",
    metricas_dt
)
resultados
````

````python
# Modelo com profundidade máxima dentro de uma variavel com um for, sem limite de folhas, sem limite de amostras mínimas por folha
profundidades = [2, 3, 5, 7, 10, 15, 20]
resultados_profundidade = []

for profundidade in profundidades:

    modelo = DecisionTreeRegressor(
        max_depth=profundidade,
        random_state=42
    )

    pipeline = Pipeline(
        steps=[
            ('preprocessor', preprocessor),
            ('modelo', modelo)
        ]
    )

    pipeline.fit(X_train, y_train)

    pred_train = pipeline.predict(X_train)
    pred_test = pipeline.predict(X_test)

    metricas_train = calcular_metricas_regressao(
        y_train,
        pred_train
    )

    metricas_test = calcular_metricas_regressao(
        y_test,
        pred_test
    )

    resultados_profundidade.append({
        'max_depth': profundidade,

        'MAE treino': metricas_train['MAE'],
        'MAE teste': metricas_test['MAE'],

        'RMSE treino': metricas_train['RMSE'],
        'RMSE teste': metricas_test['RMSE'],

        'R2 treino': metricas_train['R2'],
        'R2 teste': metricas_test['R2']
    })
````

````python
# o trade-off entre viés e variância
df_profundidade = pd.DataFrame(resultados_profundidade)

df_profundidade
````

* Cross-Validation

````python
# Importação de bibliotecas cross validation
from sklearn.model_selection import cross_validate

modelo_dt_cv = DecisionTreeRegressor(
    max_depth=10,
    random_state=42
)

pipeline_dt_cv = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('modelo', modelo_dt_cv)
    ]
)


# Validação cruzada com 5 folds, utilizando as métricas MAE, RMSE e R², retornando também o score de treino 

cv_resultados = cross_validate(
    pipeline_dt_cv,
    X_train,
    y_train,
    cv=5,
    scoring=[
        'neg_mean_absolute_error',
        'neg_root_mean_squared_error',
        'r2'
    ],
    return_train_score=True
)
````

````python
# Resultados da validação cruzada
# cv_resultados.keys()
# cv_resultados.items()

cv_df = pd.DataFrame({
    'Fold': range(1, 6),

    'MAE treino':
        -cv_resultados['train_neg_mean_absolute_error'],

    'MAE validação':
        -cv_resultados['test_neg_mean_absolute_error'],

    'RMSE treino':
        -cv_resultados['train_neg_root_mean_squared_error'],

    'RMSE validação':
        -cv_resultados['test_neg_root_mean_squared_error'],

    'R2 treino':
        cv_resultados['train_r2'],

    'R2 validação':
        cv_resultados['test_r2']
})

cv_df
````

````python
print("cv_df mean".upper())
print(cv_df.mean(numeric_only=True))
print("___________________________________________________________")
print("cv_df std".upper())
print(cv_df.std(numeric_only=True))
````

````python
# Na base de teste.
profundidades = [2, 3, 5, 7, 10, 15, 20]
resultados_cv_depth = []
for profundidade in profundidades:

    modelo = DecisionTreeRegressor(
        max_depth=profundidade,
        random_state=42
    )

    pipeline = Pipeline(
        steps=[
            ('preprocessor', preprocessor),
            ('modelo', modelo)
        ]
    )

    cv = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=5,
        scoring=[
            'neg_mean_absolute_error',
            'neg_root_mean_squared_error',
            'r2'
        ],
        return_train_score=True
    )

    resultados_cv_depth.append({
        'max_depth': profundidade,

        'MAE treino':
            -cv['train_neg_mean_absolute_error'].mean(),

        'MAE validação':
            -cv['test_neg_mean_absolute_error'].mean(),

        'RMSE treino':
            -cv['train_neg_root_mean_squared_error'].mean(),

        'RMSE validação':
            -cv['test_neg_root_mean_squared_error'].mean(),

        'R2 treino':
            cv['train_r2'].mean(),

        'R2 validação':
            cv['test_r2'].mean(),

        'R2 validação std':
            cv['test_r2'].std()
    })
    
````

##### Treinamento de modelo para 99% das transações

````python
# Limite superior para Sales no percentil 99
limite_sales_p99 = X_train['Sales'].quantile(0.99)

limite_sales_p99
````

````python
# Análise das faixas de Sales
analise_faixas_rf = pd.DataFrame({
    'Sales': X_test['Sales'],
    'Profit_real': y_test,
    'Profit_previsto': y_pred_rf_final
})

analise_faixas_rf['Faixa_Sales'] = np.where(
    analise_faixas_rf['Sales'] <= limite_sales_p99,
    'Sales <= P99',
    'Sales > P99'
)
````

````python
# Análise das métricas de regressão por faixa de Sales
resultados_faixas = []

for faixa, dados in analise_faixas_rf.groupby('Faixa_Sales'):

    metricas = calcular_metricas_regressao(
        dados['Profit_real'],
        dados['Profit_previsto']
    )

    resultados_faixas.append({
        'Faixa': faixa,
        'Quantidade': len(dados),
        **metricas
    })

resultado_faixas_rf = pd.DataFrame(resultados_faixas)

resultado_faixas_rf
````

 **Analise do Modelo**
1. O R² = 0,838 indica que, dentro desse subconjunto específico do teste, o modelo apresenta uma redução de aproximadamente 83,8% no erro quadrático em relação ao preditor que usa a média do Profit desse próprio grupo como referência. O 0,838 é uma **métrica diagnóstica do segmento Sales <= P99**
2. Agora temos evidência bastante forte de que **o desempenho do modelo se deteriora drasticamente na cauda superior de Sales.**
