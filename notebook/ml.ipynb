#### Importação de bibliotecas scikit-learn > train_test_split

```python
from sklearn.model_selection import train_test_split
```

```python
# 
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

#### Criação do One_hot_encoder

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

features_numericas = [
    'Sales',
    'Quantity',
    'Discount'
]

features_categoricas = [
    'Ship Mode',
    'Segment',
    'Region',
    'Category',
    'Sub-Category'
]

encoder = OneHotEncoder(
    handle_unknown='ignore'
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            'cat',
            encoder,
            features_categoricas
        )
    ],
    remainder='passthrough'
)
```

```python
preprocessor
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

modelo_lr = LinearRegression()

pipeline_lr = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('modelo', modelo_lr)
    ]
)

pipeline_lr.fit(X_train, y_train)

y_pred_lr = pipeline_lr.predict(X_test)
```

```python
print(y_pred_lr[:10])
```

```python
comparacao = pd.DataFrame({
    'Profit Real': y_test.values,
    'Profit Previsto': y_pred_lr
})

comparacao.head(10)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae_lr = mean_absolute_error(y_test, y_pred_lr)
mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)

print(f"MAE:  {mae_lr:.2f}")
print(f"RMSE: {rmse_lr:.2f}")
print(f"R²:   {r2_lr:.4f}")

print()
print(f"Média Profit teste:   {y_test.mean():.2f}")
print(f"Mediana Profit teste: {y_test.median():.2f}")
print(f"Desvio padrão:        {y_test.std():.2f}")
```

#### Arvore de decisão

```python
# Importação da biblioteca DecisionTreeRegressor do scikit-learn
from sklearn.tree import DecisionTreeRegressor
```

```python
# modelo sem poda explicita, avaliação de overfitting, sem regularização, sem poda, sem limite de profundidade, sem limite de folhas, sem limite de amostras mínimas por folha
modelo_dt = DecisionTreeRegressor(random_state=42)

#Criação do pipeline para o modelo de árvore de decisão
pipeline_dt = Pipeline(
    steps=[
        ('processor', preprocessor),
        ('modelo', modelo_dt)
    ]
)
```

```python
# Treinamento do modelo de árvore de decisão
pipeline_dt.fit(X_train, y_train)
```

```python
# Previsão dos valores de Profit para o conjunto de teste usando o modelo de árvore de decisão
y_pred_dt_train = pipeline_dt.predict(X_train)
y_pred_dt = pipeline_dt.predict(X_test)
# Métricas
metricas_dt_train = calcular_metricas_regressao(
    y_train,
    y_pred_dt_train
)
resultados = registrar_resultado(
    resultados,
    "Decision Tree _ train",
    metricas_dt_train
)
metricas_dt = calcular_metricas_regressao(
    y_test,
    y_pred_dt
)
resultados = registrar_resultado(
    resultados,
    "Decision Tree test",
    metricas_dt
)
resultados
```

```python
# Modelo com profundidade máxima dentro de uma variavel com um for, sem limite de folhas, sem limite de amostras mínimas por folha
profundidades = [2, 3, 5, 7, 10, 15, 20]
resultados_profundidade = []

for profundidade in profundidades:

    modelo = DecisionTreeRegressor(
        max_depth=profundidade,
        random_state=42
    )

    pipeline = Pipeline(
        steps=[
            ('preprocessor', preprocessor),
            ('modelo', modelo)
        ]
    )

    pipeline.fit(X_train, y_train)

    pred_train = pipeline.predict(X_train)
    pred_test = pipeline.predict(X_test)

    metricas_train = calcular_metricas_regressao(
        y_train,
        pred_train
    )

    metricas_test = calcular_metricas_regressao(
        y_test,
        pred_test
    )

    resultados_profundidade.append({
        'max_depth': profundidade,

        'MAE treino': metricas_train['MAE'],
        'MAE teste': metricas_test['MAE'],

        'RMSE treino': metricas_train['RMSE'],
        'RMSE teste': metricas_test['RMSE'],

        'R2 treino': metricas_train['R2'],
        'R2 teste': metricas_test['R2']
    })
```

```python
# o trade-off entre viés e variância
df_profundidade = pd.DataFrame(resultados_profundidade)

df_profundidade
```

* Cross-Validation

```python
# Importação de bibliotecas cross validation
from sklearn.model_selection import cross_validate

modelo_dt_cv = DecisionTreeRegressor(
    max_depth=10,
    random_state=42
)

pipeline_dt_cv = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('modelo', modelo_dt_cv)
    ]
)


# Validação cruzada com 5 folds, utilizando as métricas MAE, RMSE e R², retornando também o score de treino 

cv_resultados = cross_validate(
    pipeline_dt_cv,
    X_train,
    y_train,
    cv=5,
    scoring=[
        'neg_mean_absolute_error',
        'neg_root_mean_squared_error',
        'r2'
    ],
    return_train_score=True
)
```

```python
# Resultados da validação cruzada
# cv_resultados.keys()
# cv_resultados.items()

cv_df = pd.DataFrame({
    'Fold': range(1, 6),

    'MAE treino':
        -cv_resultados['train_neg_mean_absolute_error'],

    'MAE validação':
        -cv_resultados['test_neg_mean_absolute_error'],

    'RMSE treino':
        -cv_resultados['train_neg_root_mean_squared_error'],

    'RMSE validação':
        -cv_resultados['test_neg_root_mean_squared_error'],

    'R2 treino':
        cv_resultados['train_r2'],

    'R2 validação':
        cv_resultados['test_r2']
})

cv_df
```

```python
print("cv_df mean".upper())
print(cv_df.mean(numeric_only=True))
print("___________________________________________________________")
print("cv_df std".upper())
print(cv_df.std(numeric_only=True))
```

```python
# Na base de teste.
profundidades = [2, 3, 5, 7, 10, 15, 20]
resultados_cv_depth = []
for profundidade in profundidades:

    modelo = DecisionTreeRegressor(
        max_depth=profundidade,
        random_state=42
    )

    pipeline = Pipeline(
        steps=[
            ('preprocessor', preprocessor),
            ('modelo', modelo)
        ]
    )

    cv = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=5,
        scoring=[
            'neg_mean_absolute_error',
            'neg_root_mean_squared_error',
            'r2'
        ],
        return_train_score=True
    )

    resultados_cv_depth.append({
        'max_depth': profundidade,

        'MAE treino':
            -cv['train_neg_mean_absolute_error'].mean(),

        'MAE validação':
            -cv['test_neg_mean_absolute_error'].mean(),

        'RMSE treino':
            -cv['train_neg_root_mean_squared_error'].mean(),

        'RMSE validação':
            -cv['test_neg_root_mean_squared_error'].mean(),

        'R2 treino':
            cv['train_r2'].mean(),

        'R2 validação':
            cv['test_r2'].mean(),

        'R2 validação std':
            cv['test_r2'].std()
    })
    
```

```python
df_cv_depth = pd.DataFrame(resultados_cv_depth)

df_cv_depth
```

* GridSearchCV


```python
# Importação da biblioteca GridSearchCV do scikit-learn
from sklearn.model_selection import GridSearchCV
```

```python
# Definição do modelo de árvore de decisão

modelo_dt_grid = DecisionTreeRegressor(
    random_state=42
)

pipeline_dt_grid = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('modelo', modelo_dt_grid)
    ]
)

# Definição dos hiperparâmetros a serem testados no GridSearchCV
param_grid = {
    'modelo__max_depth': [5, 10, 15, 20, None],
    'modelo__min_samples_split': [2, 5, 10],
    'modelo__min_samples_leaf': [1, 2, 5, 10]
}
```

```python
# Criando o objeto GridSearchCV
grid_dt = GridSearchCV(
    estimator=pipeline_dt_grid,
    param_grid=param_grid,
    scoring='neg_root_mean_squared_error',
    cv=5,
    n_jobs=-1,
    return_train_score=True
)

# Estimador
estimator=pipeline_dt_grid

# param_grid
param_grid=param_grid

# Scoring
scoring='neg_root_mean_squared_error'
```

```python
# Treinando o modelo com GridSearchCV
grid_dt.fit(
    X_train,
    y_train
)
```

```python
# Comparando combinações de hiperparâmetros
print(grid_dt.best_params_)
print("____________________________________________________________")
print(-grid_dt.best_score_)
print("____________________________________________________________")
melhor_dt = grid_dt.best_estimator_
print(melhor_dt)
# print("____________________________________________________________")
# print(melhor_dt.predict(X_test))
```

```python
# Refinamento do modelo de árvore de decisão com os melhores hiperparâmetros encontrados pelo GridSearchCV
param_grid_refinado = {
    'modelo__max_depth': [15, 20, 25, 30, None],
    'modelo__min_samples_split': [2, 3, 5],
    'modelo__min_samples_leaf': [1, 2, 3]
}

grid_dt_refinado = GridSearchCV(
    estimator=pipeline_dt_grid,
    param_grid=param_grid_refinado,
    scoring='neg_root_mean_squared_error',
    cv=5,
    n_jobs=-1,
    return_train_score=True
)
```

```python
# Treinando o modelo refinado com GridSearchCV
grid_dt_refinado.fit(
    X_train,
    y_train
)
```

```python
# Obtenção dos melhores hiperparâmetros do modelo refinado
print(grid_dt_refinado.best_params_)
print("____________________________________________________________")
print(-grid_dt_refinado.best_score_)
```

```python
# Recuperando o melhor modelo refinado
melhor_dt = grid_dt_refinado.best_estimator_

# Validando o modelo refinado na base de teste
y_pred_dt_final = melhor_dt.predict(X_test)

# Métricas do modelo refinado
metricas_dt_final = calcular_metricas_regressao(
    y_test,
    y_pred_dt_final
)

metricas_dt_final
```

```python
# Registrando os resultados do modelo refinado
resultados = registrar_resultado(
    resultados,
    'Decision Tree Otimizada',
    metricas_dt_final
)
resultados
```

##### Analise de Residuos e erros!

```python
analise_erros_dt = pd.DataFrame({
    'Profit_real': y_test,
    'Profit_previsto': y_pred_dt_final
})
analise_erros_dt['Erro'] = (
    analise_erros_dt['Profit_real']
    - analise_erros_dt['Profit_previsto']
)
analise_erros_dt['Erro_absoluto'] = (
    analise_erros_dt['Erro'].abs()
)

maiores_erros_dt = (
    analise_erros_dt
    .sort_values(
        'Erro_absoluto',
        ascending=False
    )
    .head(20)
)
Erro = maiores_erros_dt['Profit_real'] - maiores_erros_dt['Profit_previsto']
print("Erros")
Erro
print("Maiores erros do modelo Decision Tree Otimizada:")
maiores_erros_dt
```

```python
casos_extremos_dt = casos_extremos_dt.sort_values(
    'Erro_absoluto',
    ascending=False
)

casos_extremos_dt
```

#### Experimento de diagnostico das variações do erro quadratico RMSE


```python
analise_erros_dt['Erro_quadrado'] = (
    analise_erros_dt['Erro'] ** 2
)
erro_quadrado_total = (
    analise_erros_dt['Erro_quadrado'].sum()
)
top20_erro_quadrado = (
    analise_erros_dt
    .nlargest(20, 'Erro_absoluto')['Erro_quadrado']
    .sum()
)
percentual_top20 = (
    top20_erro_quadrado
    / erro_quadrado_total
) * 100

percentual_top20
```
